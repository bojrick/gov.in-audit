```python
import pandas as pd
from bs4 import BeautifulSoup
import requests

url = 'https://igod.gov.in/site_map'
# Assuming you have already defined the 'url' variable with the target website URL
response = requests.get(url)
soup = BeautifulSoup(response.text, 'html.parser')

# Extract all links and their text
links = soup.find_all('a')
data = []
for link in links:
    href = link.get('href')
    text = link.text.strip()
    data.append({'Link': href, 'Text': text})

# Create a pandas DataFrame
df = pd.DataFrame(data)

# Display the first few rows of the DataFrame
print(df.head())
```

```python
# Filter the DataFrame to keep only links containing '.in'
df_filtered = df[df['Link'].str.contains('.in', na=False)]
# df_filtered = df_filtered[~df_filtered['Link'].str.contains('igod')]

# Display the filtered DataFrame
df_filtered
```
